```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 17000, 22000, 25000]
}

df = pd.DataFrame(data)

# Outputs
print("Sales Data")
print(df)

print("\nTotal Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())

highest_month = df.loc[df["Sales"].idxmax(), "Month"]
lowest_month = df.loc[df["Sales"].idxmin(), "Month"]

print("Highest Sales Month:", highest_month)
print("Lowest Sales Month:", lowest_month)

df["Growth"] = df["Sales"].pct_change() * 100
print("\nMonthly Growth (%)")
print(df[["Month", "Growth"]])

# Graph
plt.figure(figsize=(8,5))
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Sales Trend Visualization")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# Median Sales
print("Median Sales:", df["Sales"].median())

# Sales Range
sales_range = df["Sales"].max() - df["Sales"].min()
print("Sales Range:", sales_range)

# Standard Deviation
print("Sales Standard Deviation:", round(df["Sales"].std(), 2))
```
